### In Synthetic data

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
````

````python
#Crear grafico de dispersion de las muestras
X, y = make_classification(n_samples = 100 , n_features = 2 , n_redundant = 0, n_informative = 2, n_clusters_per_class = 1, n_classes = 2, random_state = 1)
#print (X.shape, y.shape, X.T) x.t es la traspuesta

fig, ax = plt.subplots(figsize=( 6 , 5 )) #Controla size
ax.scatter( X[:, 0] , X[:, 1] , c = y , cmap = 'Paired') # c el var de colores
ax.set_title('Synthetic data')
plt.show()
````

````python
# Crear y entrenar el clasificador árbol de decisión
# Create an instance
plt.show()
decision_tree = DecisionTreeClassifier()
#Encontrar los panarametros segun los datos que tenemos se usa fit para entrenar (var ind = datos (los datos deben de tener forma especifica)) siempre las muestras van por las filas
# Train the model
decision_tree.fit ( X , y ) #(# Samples, # Features)
````

````python
# predicciones para ajustar el modelo
predictions = decision_tree.predict(X)
print (predictions)
````

### **Decision boundaries**

````python
# Definir rango para el grid sobre el cual visualizar la frontera
# Min-Max values for the grid of decision boundaries
#x_min, x_max = 0 , 1
#y_min, y_max = 0 , 1
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

# Create grid
#x_grid, y_grid = np.meshgrid( X , y )
x_grid, y_grid = np.meshgrid(
    np.linspace(x_min, x_max, 100),
    np.linspace(y_min, y_max, 100)
)
# Estimate predictions over the grid
predictions = decision_tree.predict(np.c_[x_grid.ravel(), y_grid.ravel()])
print ("Shape of predictions" , predictions.shape)
````

````python
# Reshape predictions
predictions = predictions.reshape(x_grid.shape)

# Draw grid
fig, ax = plt.subplots(figsize=(6, 5))
ax.pcolormesh(x_grid, y_grid, predictions, cmap='Paired', shading='auto')

# Scatter real values
# Dar forma al vector de predicciones para que coincida con el grid
ax.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap='Paired')

ax.set_xticks(())
ax.set_yticks(())
plt.show()
````

### In Real data

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.datasets import load_iris
````

````python
iris = load_iris() # Load iris dataset
iris_x = iris.data # # Features (4 características)
iris_y = iris.target # Labels (3 clases)

# Imprimir formas
print (iris_x.shape, iris_y.shape)
````

````python
# Create an instance
# (150, 4) (150,)
decision_tree = DecisionTreeClassifier()

# Train the model
decision_tree.fit ( iris_x , iris_y ) #(# Samples, # Features)
````

````python
fig, ax = plt.subplots(figsize=( 12 , 5 ))
plot_tree(decision_tree, filled=True, fontsize=5, ax=ax)
plt.title("Árbol de Decisión entrenado con el dataset Iris")
plt.show()
````
