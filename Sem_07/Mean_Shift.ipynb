# Mean Shift - Generated data

Mean shift documentation [here](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.MeanShift.html)

```python
import numpy as np
import pandas as pd
from sklearn.cluster import MeanShift
from sklearn.datasets import make_blobs
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split #funcion permite dividir los datos
```

```python
n_samples =  400
random_state = 3
X, _ = make_blobs(n_samples = n_samples, n_features=2, random_state = random_state) #aprendizaje no supervisado  no se usan los labels

x_train, x_test = train_test_split(X, test_size=0.2, random_state=3)

_, axes = plt.subplots(1,2, figsize=(13,5))
axes[0].scatter( X[:, 0] , X[:, 1] , edgecolors='k')
axes[1].scatter( x_train[:, 0] , x_train[:, 1] , edgecolors='k', cmap= 'Paired')
axes[1].scatter( x_test[:, 0] , x_test[:, 1] , edgecolors='k', cmap= 'Paired')
```

```python
ms = MeanShift( bandwidth= 2.72 ) #se halla sumando los valores del eje Y y X (15% aprox. se calculo para este ejemplo)
ms.fit(x_train)
```

```python
cluster_centers = ms.cluster_centers_

predictions = ms.predict(x_test)

_, ax = plt.subplots(figsize=(6,5))
ax.scatter(x_test[:, 0] , x_test[:, 1] , c = predictions,edgecolors='k', cmap= 'Paired')
ax.scatter(cluster_centers[:,0], cluster_centers[:,1], edgecolors='k', marker='*', s = 500)
```
