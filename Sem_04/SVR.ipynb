## In Synthetic data

````python
import numpy as np # Librería mateamtica
import matplotlib.pyplot as plt # plots

from sklearn.svm import SVR
from sklearn.datasets import make_regression
````

````python
X, y = make_regression(n_samples = 200, n_features = 2, noise = 10, random_state = 1)

_, ax = plt.subplots (figsize=(6,5))
ax.scatter( X[:, 1] , y , c = 'b') #usar 0 y 1 para crear la relacion con y
````

````python
svr = SVR(kernel = 'sigmoid' )

svr.fit( X , y )
````

````python
# Predecir la nueva muestra. Imprimir la prediccion
prediction = svr.predict( X )
# Predict x values
print (prediction.shape)
````

````python
# Draw real samples (blue) and predictions (red)
_, axes = plt.subplots (1,2, figsize = ( 10 , 5 )) #1 fil 2 cols
axes[0].scatter( X[:, 0] , y , color = 'blue', alpha=0.5)
axes[0].scatter ( X[:, 0] , prediction , color = 'red')
axes[1].scatter( X[:, 1] , y , color = 'blue', alpha=0.5)
axes[1].scatter ( X[:, 1] , prediction , color = 'red')
````
