# Let's start on the basics. Variables and arrays

### First, import some libraries

```
import numpy as np
```

```
array_1 = np.array([1, 2, 8, 9, 5, 5])        # Create an array of N values
array_2 = np.array([10, 20, 80, 90, 50, 50])  # Create an array of random int values
                                              # with the same lenght as array_1

print(array_1, array_2) # print both array_1 and array_2
print(array_1.shape, array_2.shape) # print both array_1 and array_2 shapes
print(array_1 + array_2) # print summation of array_1 and array_2
```

```
array_rp = array_1[3] = 70 # Replace the value of array_1 in pos 3 by 70
print(array_1) # Print summation again
```

```
array_mlt = array_1 * array_2 # Apply element-wise multiplication among array_1 and array_2
print(array_1, array_2)
print(array_mlt) # Print results
```

## Load an image using OpenCV

```
# Loading a color image
img = cv2.imread( "/content/lena.jpg" )
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # set the color

print(img.shape) #dimensions h*l and color
_, ax = plt.subplots(figsize=(6, 4))
ax.imshow(img) # plots an image
```

### Let's play with the data

```
# Printing the image shape (size of the image)
print(img.shape) #dimensions h*l and color
img = img.astype(np.uint8)
img[0:100, 0:20, :] = 255
_, ax = plt.subplots(figsize=(6,4))
ax.imshow(img)
```

```
# Looking through the image...
# print among image indexes
print(img[100:110, 100:110])
```

```
# Changing its channels values!
img = cv2.imread("/content/lena.jpg")
img[:, :, 0] = 0 # Set the red channel to 0
plt.imshow(img) # Voila!
```
