# **K-means - isotropic data**

```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
```

```
# Generate data
X, _ = make_blobs(n_samples = 200 , n_features = 2, random_state = 3)

# Create train / test splits
X_train, X_test = train_test_split(X, test_size = 0.2, random_state= 3)

# Plot everything and print all shapes
_, axes = plt.subplots(1, 2, figsize=(10,4))
axes[0].scatter( X[:, 0] , X[:, 1] , edgecolors='k')
axes[1].scatter( X_train[:, 0] , X_train[:, 1], c = 'g' , edgecolors='k')
axes[1].scatter( X_test[:, 0] , X_test[:, 1], c = 'r' , edgecolors='k')

print (X.shape, X_train.shape, X_test.shape)
```

## **Elbow method**

```
# Create a distance function
def euclidean_distance(pt1, pt2): # metrica de distancia entre dos puntos
  assert len(pt1) == 2, "Error! No 2d point" #validacion como un true inicialmente detiene la ejecucion automaticamente si entra a False - solo funciona con 2 dimensiones
  assert len(pt2) == 2, "Error! No 2d point"
  return (pt1[0] - pt2[0]) ** 2 + (pt1[1] - pt2[1]) ** 2 # formula de la distancia euclidia
```

```
# Define num of clusters
clusters = [ 2, 3, 4, 5, 6, 7, 8, 9 ] #inicializa cantidad de clusters
elbow_distances = [] # guarda distancia de cada cluster

# Eval kmeans over each num of clusters
for cluster in clusters:
  kmeans = KMeans(n_clusters = cluster ) #definicion de algoritmos
  kmeans.fit( X_train )

  centroides = kmeans.cluster_centers_ #sacar centroides y medir cada muestra de cada con respecto al centroide
  predictions = kmeans.predict( X_test )
  distance = 0

  for i in range(len(X_test)) : #numero de muestras de test
    centroide = centroides[predictions[i]] #aqui se tendria el centroide mas cercano
    distance += euclidean_distance( centroide , X_test[i] ) #calculo del centroide con respecto ala muestra

  elbow_distances.append(distance/len(X_test))

print (len(elbow_distances))
```

```
# Plot the elbow
_, ax = plt.subplots (figsize = (6,5)) # siempre el menor numero de centroides, grupos posibles en donde esta el elbow se dice que es en donde tenemos los clusters (k)
ax.plot( clusters , elbow_distances , marker="x") #menor grupos posibles vs mayor grupos posibles
ax.set_title("Elbow") #cambio brusco en distancia con respecto al anterior y que en el sgte punto no haya un cambio significativo
```

## **Silhouette method**

```
from sklearn.metrics import silhouette_score
# metrica de distancias entre las muestras de cd grupo y considera la distancia interclass
# (mirar que tanto se separan los grupos (score de la silueta es grande -1: dispersos y juntos todos los centroides y 0: las muestras estan dispersas pero los grupos estan separados (se puede hacer agrupamiento) y 1: grupos compactos y sepados (mejor resultado)))
# Define num of clusters
clusters = [ 2, 3, 4, 5, 6, 7, 8, 9 ]
silhouette_distances = []

# Eval kmeans over each num of clusters
for cluster in clusters :
  kmeans = KMeans(n_clusters = cluster )
  kmeans.fit( X_train )

  predictions = kmeans.predict( X_test )
  distance = silhouette_score( X_test , predictions , metric="euclidean")

  silhouette_distances.append(distance) #guardarla en la lista

print (len(silhouette_distances))
```

```
# Plot the silhouette
_, ax = plt.subplots (figsize = (6,5))
ax.plot( clusters , silhouette_distances , marker="x")
ax.set_title("Silhouette")
```
