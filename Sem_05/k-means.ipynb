# K Means - Generated data

Isotropic dataset module in sklearn [here](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_blobs.html) \
Train-test splits in sklearn [here](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html) \
k-Means module in sklearn [here](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html)

```
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
```

```
X, _ = make_blobs(n_samples = 250, n_features = 2 , random_state = 170 ) # _ forma de optimizar memoria

# Create train (80%) and test (20%) sets and print all shapes
#dividir datos
x_train, x_test = train_test_split(X, test_size = 0.2 , random_state = 170)
print(x_train.shape)
print(x_test.shape)
_, axes = plt.subplots(1,2, figsize=(13,5))
axes[0].scatter( X[:, 0] , X[:, 1] , edgecolors='k')
axes[1].scatter( x_train[:, 0] , x_train[:, 1] , edgecolors='g' , label='Train')
axes[1].scatter( x_test[:, 0] , x_test[:, 1] , edgecolors='r' , label='Test')
axes[1].legend()
```

```
kmeans = KMeans(n_clusters=3) #objeto que se encarga de hacer el agrupamiento - 3 centroides o mediodes

# Train Kmeans
kmeans.fit(x_train)
```

```
# Plot centroids
centroids = kmeans.cluster_centers_ # indica en donde se ubican los centriodes
print(centroids)
```

```
_, ax = plt.subplots (figsize = (15,5))
colors=['blue','cyan','magenta'] # azul test

# plot data and found centroids
ax.scatter(x_test[:, 0], x_test[:, 1], edgecolors ='k' )
ax.scatter(centroids[:, 0], centroids[:, 1], c = colors, marker='*', s = 700, cmap='viridis')  #cambiar vineta rombos, estrellas... (marker) y tamano (s) de los puntos
```

```
# Plot decision boundaries
Predictions = kmeans.predict(x_test)
```
